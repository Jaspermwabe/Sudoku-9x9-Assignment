```python
"""
Advanced Algorithm - Assignment 2
=================================
1. Check whether a puzzle is a standard Sudoku puzzle.
2. Solve it using
   (a) the Backtracking method, and
   (b) the Constraint Propagation method (elimination + naked/hidden singles,
       with a depth-first search that uses the MRV heuristic when propagation
       alone cannot finish).

Run:  python sudoku_assignment2.py
Only the Python standard library is used.
"""

import time

# --------------------------------------------------------------------------
# Puzzle from the assignment sheet (0 = empty cell)
# --------------------------------------------------------------------------
PUZZLE = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9],
]

SIZE = 9
BOX = 3
DIGITS = range(1, SIZE + 1)

# --------------------------------------------------------------------------
# Shared helpers: units (rows, columns, boxes) and peers
# --------------------------------------------------------------------------
CELLS = [(r, c) for r in range(SIZE) for c in range(SIZE)]

ROW_UNITS = [[(r, c) for c in range(SIZE)] for r in range(SIZE)]
COL_UNITS = [[(r, c) for r in range(SIZE)] for c in range(SIZE)]
BOX_UNITS = [
    [(r, c)
     for r in range(br, br + BOX)
     for c in range(bc, bc + BOX)]
    for br in range(0, SIZE, BOX)
    for bc in range(0, SIZE, BOX)
]
ALL_UNITS = ROW_UNITS + COL_UNITS + BOX_UNITS

# UNITS_OF[cell] -> the 3 units (row, column, box) containing the cell
UNITS_OF = {cell: [u for u in ALL_UNITS if cell in u] for cell in CELLS}
# PEERS[cell] -> the 20 other cells that share a unit with it
PEERS = {
    cell: set(c for u in UNITS_OF[cell] for c in u) - {cell}
    for cell in CELLS
}


def copy_grid(grid):
    return [row[:] for row in grid]


def print_grid(grid, title=None):
    """Pretty-print a 9x9 grid (0 is shown as '.')."""
    if title:
        print(title)
    for r in range(SIZE):
        if r % BOX == 0 and r != 0:
            print("------+-------+------")
        parts = []
        for c in range(SIZE):
            if c % BOX == 0 and c != 0:
                parts.append("|")
            parts.append(str(grid[r][c]) if grid[r][c] else ".")
        print(" ".join(parts))
    print()


def candidates(grid, r, c):
    """Digits that can legally be placed at (r, c) given the current grid."""
    used = set(grid[r])
    used.update(grid[i][c] for i in range(SIZE))
    br, bc = (r // BOX) * BOX, (c // BOX) * BOX
    used.update(grid[i][j]
                for i in range(br, br + BOX)
                for j in range(bc, bc + BOX))
    return [d for d in DIGITS if d not in used]


def is_valid_solution(grid):
    """True if grid is completely filled and every unit holds 1..9 once."""
    full = set(DIGITS)
    return all(
        set(grid[r][c] for r, c in unit) == full for unit in ALL_UNITS
    )


# --------------------------------------------------------------------------
# TASK 1: Is it a standard Sudoku puzzle?
# --------------------------------------------------------------------------
def count_solutions(grid, limit=2):
    """
    Count solutions, stopping once `limit` have been found.
    Uses backtracking with the fewest-candidates-first rule so that the
    uniqueness test stays fast.
    """
    g = copy_grid(grid)
    count = 0

    def search():
        nonlocal count
        best, best_opts = None, None
        for r, c in CELLS:
            if g[r][c] == 0:
                opts = candidates(g, r, c)
                if best is None or len(opts) < len(best_opts):
                    best, best_opts = (r, c), opts
                    if len(opts) <= 1:
                        break
        if best is None:                       # no empty cell: a solution
            count += 1
            return
        r, c = best
        for d in best_opts:
            g[r][c] = d
            search()
            g[r][c] = 0
            if count >= limit:
                return

    search()
    return count


def check_standard_sudoku(grid):
    """
    A standard Sudoku puzzle must satisfy ALL of these:
      1. Shape      : exactly 9 rows x 9 columns.
      2. Values     : every entry is an integer 0..9 (0 = empty).
      3. Consistent : no digit repeats in any row, column or 3x3 box.
      4. Solvable   : at least one completion exists.
      5. Unique     : exactly one completion exists (well-formed puzzle).
    Returns (is_standard, list_of_messages).
    """
    msgs = []

    # 1. Shape
    if not isinstance(grid, (list, tuple)) or len(grid) != SIZE or \
            any(not isinstance(row, (list, tuple)) or len(row) != SIZE
                for row in grid):
        return False, ["Grid is not 9 x 9."]

    # 2. Values
    for r in range(SIZE):
        for c in range(SIZE):
            v = grid[r][c]
            if isinstance(v, bool) or not isinstance(v, int) or not 0 <= v <= 9:
                return False, [f"Invalid value {v!r} at row {r + 1}, "
                               f"column {c + 1} (must be an integer 0-9)."]

    # 3. Consistency of the given clues
    names = (["row %d" % (i + 1) for i in range(SIZE)]
             + ["column %d" % (i + 1) for i in range(SIZE)]
             + ["box %d" % (i + 1) for i in range(SIZE)])
    for name, unit in zip(names, ALL_UNITS):
        seen = set()
        for r, c in unit:
            v = grid[r][c]
            if v == 0:
                continue
            if v in seen:
                msgs.append(f"Duplicate {v} in {name}.")
            seen.add(v)
    if msgs:
        return False, msgs

    # 4 & 5. Solvable and unique
    n = count_solutions(grid, limit=2)
    if n == 0:
        return False, ["Puzzle has no solution."]
    if n > 1:
        return False, ["Puzzle has more than one solution (not unique)."]

    clues = sum(1 for r, c in CELLS if grid[r][c])
    return True, [f"9 x 9 grid, {clues} clues, no conflicts, exactly one solution."]


# --------------------------------------------------------------------------
# TASK 2(a): Backtracking
# --------------------------------------------------------------------------
def solve_backtracking(grid):
    """
    Plain depth-first backtracking:
      find the next empty cell -> try 1..9 -> keep a digit only if it breaks no
      rule -> recurse -> undo (backtrack) when the branch dead-ends.
    Returns (solution or None, stats dict).
    """
    g = copy_grid(grid)
    stats = {"calls": 0, "backtracks": 0}

    def solve():
        stats["calls"] += 1
        for r, c in CELLS:                       # next empty cell
            if g[r][c] == 0:
                for d in candidates(g, r, c):    # try digits that fit
                    g[r][c] = d
                    if solve():
                        return True
                    g[r][c] = 0                  # undo
                    stats["backtracks"] += 1
                return False                     # dead end
        return True                              # no empty cell left

    ok = solve()
    return (g if ok else None), stats


# --------------------------------------------------------------------------
# TASK 2(b): Constraint propagation (+ search with MRV)
# --------------------------------------------------------------------------
# State: dict  cell -> set of digits still possible in that cell.
# Two rules are applied repeatedly until nothing changes:
#   Elimination / naked single : if a cell has one candidate left, remove that
#                                digit from all 20 peers.
#   Hidden single              : if a digit fits only one cell of a unit,
#                                place it there.

def assign(cands, cell, d, stats):
    """Fix `cell` to `d` by eliminating all other digits. False = contradiction."""
    for other in list(cands[cell] - {d}):
        if not eliminate(cands, cell, other, stats):
            return False
    return True


def eliminate(cands, cell, d, stats):
    """Remove d from cands[cell] and propagate. False = contradiction."""
    if d not in cands[cell]:
        return True                              # already done
    cands[cell].discard(d)
    stats["eliminations"] += 1

    if not cands[cell]:                          # no digit left -> contradiction
        return False

    if len(cands[cell]) == 1:                    # rule 1: naked single
        only = next(iter(cands[cell]))
        for p in PEERS[cell]:
            if not eliminate(cands, p, only, stats):
                return False

    for unit in UNITS_OF[cell]:                  # rule 2: hidden single
        places = [s for s in unit if d in cands[s]]
        if not places:
            return False                         # digit has nowhere to go
        if len(places) == 1:
            if not assign(cands, places[0], d, stats):
                return False
    return True


def init_candidates(grid, stats):
    """Build the candidate table and propagate all given clues."""
    cands = {cell: set(DIGITS) for cell in CELLS}
    for r, c in CELLS:
        if grid[r][c] and not assign(cands, (r, c), grid[r][c], stats):
            return None                          # clues contradict each other
    return cands


def cands_to_grid(cands):
    return [[next(iter(cands[(r, c)])) if len(cands[(r, c)]) == 1 else 0
             for c in range(SIZE)] for r in range(SIZE)]


def solve_constraint_propagation(grid):
    """
    Constraint propagation first; if cells remain undecided, depth-first search
    on the cell with the fewest candidates (MRV), propagating after each guess.
    Returns (solution or None, stats dict).
    """
    stats = {"eliminations": 0, "guesses": 0, "solved_by_propagation_alone": False}

    cands = init_candidates(grid, stats)
    if cands is None:
        return None, stats

    if all(len(v) == 1 for v in cands.values()):
        stats["solved_by_propagation_alone"] = True
        return cands_to_grid(cands), stats

    def search(state):
        if all(len(v) == 1 for v in state.values()):
            return state
        # MRV: undecided cell with the fewest candidates
        cell = min((s for s in CELLS if len(state[s]) > 1),
                   key=lambda s: len(state[s]))
        for d in sorted(state[cell]):
            stats["guesses"] += 1
            trial = {k: set(v) for k, v in state.items()}   # copy state
            if assign(trial, cell, d, stats):
                result = search(trial)
                if result:
                    return result
        return None

    final = search(cands)
    return (cands_to_grid(final) if final else None), stats


# --------------------------------------------------------------------------
# Demo / main
# --------------------------------------------------------------------------
def timed(fn, *args):
    t0 = time.perf_counter()
    out = fn(*args)
    return out, (time.perf_counter() - t0) * 1000.0     # milliseconds


def main():
    print("=" * 60)
    print("TASK 1: Is it a standard Sudoku puzzle?")
    print("=" * 60)
    print_grid(PUZZLE, "Assignment puzzle:")
    ok, msgs = check_standard_sudoku(PUZZLE)
    print("Standard Sudoku:", ok)
    for m in msgs:
        print("  -", m)

    # Extra demonstrations of the checker on bad inputs
    print("\nChecker on other inputs:")
    bad_duplicate = copy_grid(PUZZLE)
    bad_duplicate[0][2] = 5                     # second 5 in row 1
    bad_shape = [row[:8] for row in PUZZLE]     # only 8 columns
    bad_value = copy_grid(PUZZLE)
    bad_value[0][2] = 10
    empty_grid = [[0] * SIZE for _ in range(SIZE)]
    for label, g in [("duplicate digit", bad_duplicate),
                     ("wrong shape", bad_shape),
                     ("value out of range", bad_value),
                     ("empty grid", empty_grid)]:
        res, why = check_standard_sudoku(g)
        print(f"  {label:20s} -> {res}  ({why[0]})")

    print("\n" + "=" * 60)
    print("TASK 2(a): Backtracking")
    print("=" * 60)
    (sol_bt, st_bt), t_bt = timed(solve_backtracking, PUZZLE)
    print_grid(sol_bt, "Solution (backtracking):")
    print(f"Recursive calls: {st_bt['calls']:,}   "
          f"Backtracks: {st_bt['backtracks']:,}   Time: {t_bt:.2f} ms")
    print("Valid solution:", is_valid_solution(sol_bt))

    print("\n" + "=" * 60)
    print("TASK 2(b): Constraint propagation")
    print("=" * 60)
    (sol_cp, st_cp), t_cp = timed(solve_constraint_propagation, PUZZLE)
    print_grid(sol_cp, "Solution (constraint propagation):")
    print(f"Eliminations: {st_cp['eliminations']:,}   "
          f"Guesses: {st_cp['guesses']}   Time: {t_cp:.2f} ms")
    print("Solved by propagation alone (no guessing):",
          st_cp["solved_by_propagation_alone"])
    print("Valid solution:", is_valid_solution(sol_cp))

    print("\n" + "=" * 60)
    print("COMPARISON")
    print("=" * 60)
    print("Both methods give the same solution:", sol_bt == sol_cp)
    print(f"{'Method':<24}{'Work done':<34}{'Time (ms)':>10}")
    print(f"{'Backtracking':<24}{str(st_bt['calls']) + ' calls, ' + str(st_bt['backtracks']) + ' backtracks':<34}{t_bt:>10.2f}")
    print(f"{'Constraint propagation':<24}{str(st_cp['guesses']) + ' guesses, ' + str(st_cp['eliminations']) + ' eliminations':<34}{t_cp:>10.2f}")


if __name__ == "__main__":
    main()
```

```python
from copy import deepcopy
from typing import Dict, List, Optional, Set, Tuple

Grid = List[List[int]]
Coord = Tuple[int, int]
Domains = List[List[Set[int]]]


# ==========================================
# 1. Check if the Sudoku is a Standard Puzzle
# ==========================================
def is_standard_sudoku(grid: Grid) -> bool:
    """
    Validates that `grid` is a standard 9x9 Sudoku board:
    - 9 rows and 9 columns
    - Integers in range 0..9 (0 = empty, 1..9 = given clue)
    - No duplicate non-zero digits in any row, column, or 3x3 box
    """
    if not isinstance(grid, list) or len(grid) != 9:
        return False

    for row in grid:
        if not isinstance(row, list) or len(row) != 9:
            return False
        for val in row:
            if type(val) is not int or val < 0 or val > 9:
                return False

    # Check rows and columns
    for i in range(9):
        row_seen: Set[int] = set()
        col_seen: Set[int] = set()
        for j in range(9):
            r_val = grid[i][j]
            c_val = grid[j][i]

            if r_val != 0:
                if r_val in row_seen:
                    return False
                row_seen.add(r_val)

            if c_val != 0:
                if c_val in col_seen:
                    return False
                col_seen.add(c_val)

    # Check 3x3 boxes
    for box_r in range(0, 9, 3):
        for box_c in range(0, 9, 3):
            box_seen: Set[int] = set()
            for r in range(box_r, box_r + 3):
                for c in range(box_c, box_c + 3):
                    val = grid[r][c]
                    if val != 0:
                        if val in box_seen:
                            return False
                        box_seen.add(val)

    return True


# ==========================================
# 2a. Backtracking Solver
# ==========================================
def _is_valid_placement(grid: Grid, row: int, col: int, num: int) -> bool:
    for i in range(9):
        if grid[row][i] == num or grid[i][col] == num:
            return False

    box_r = (row // 3) * 3
    box_c = (col // 3) * 3
    for r in range(box_r, box_r + 3):
        for c in range(box_c, box_c + 3):
            if grid[r][c] == num:
                return False

    return True


def _find_empty(grid: Grid) -> Optional[Coord]:
    for r in range(9):
        for c in range(9):
            if grid[r][c] == 0:
                return (r, c)
    return None


def solve_backtracking(grid: Grid) -> Optional[Grid]:
    """
    Solves a 9x9 Sudoku grid using recursive backtracking.
    Returns a solved copy of the grid, or None if unsolvable.
    """
    if not is_standard_sudoku(grid):
        return None

    board = deepcopy(grid)

    def backtrack() -> bool:
        empty = _find_empty(board)
        if empty is None:
            return True

        row, col = empty
        for num in range(1, 10):
            if _is_valid_placement(board, row, col, num):
                board[row][col] = num
                if backtrack():
                    return True
                board[row][col] = 0

        return False

    return board if backtrack() else None


# ==========================================
# 2b. Constraint Propagation + Search Solver
# ==========================================
def _build_peers_map() -> Dict[Coord, Set[Coord]]:
    peers_map: Dict[Coord, Set[Coord]] = {}
    for r in range(9):
        for c in range(9):
            peers: Set[Coord] = set()
            for i in range(9):
                if i != c:
                    peers.add((r, i))
                if i != r:
                    peers.add((i, c))
            box_r, box_c = (r // 3) * 3, (c // 3) * 3
            for br in range(box_r, box_r + 3):
                for bc in range(box_c, box_c + 3):
                    if (br, bc) != (r, c):
                        peers.add((br, bc))
            peers_map[(r, c)] = peers
    return peers_map


def _get_units(row: int, col: int) -> List[List[Coord]]:
    row_unit = [(row, c) for c in range(9)]
    col_unit = [(r, col) for r in range(9)]
    box_r, box_c = (row // 3) * 3, (col // 3) * 3
    box_unit = [
        (r, c) for r in range(box_r, box_r + 3) for c in range(box_c, box_c + 3)
    ]
    return [row_unit, col_unit, box_unit]


PEERS_MAP = _build_peers_map()


def _eliminate(domains: Domains, row: int, col: int, val: int) -> bool:
    if val not in domains[row][col]:
        return True

    domains[row][col].remove(val)

    # Contradiction: no possible values left
    if len(domains[row][col]) == 0:
        return False

    # Naked single: propagate forced value to peers
    if len(domains[row][col]) == 1:
        only_val = next(iter(domains[row][col]))
        for pr, pc in PEERS_MAP[(row, col)]:
            if not _eliminate(domains, pr, pc, only_val):
                return False

    # Hidden single: if `val` has only one valid position in a unit, assign it
    for unit in _get_units(row, col):
        places = [(r, c) for (r, c) in unit if val in domains[r][c]]
        if len(places) == 0:
            return False
        if len(places) == 1:
            ur, uc = places[0]
            if len(domains[ur][uc]) > 1:
                if not _assign(domains, ur, uc, val):
                    return False

    return True


def _assign(domains: Domains, row: int, col: int, val: int) -> bool:
    if val not in domains[row][col]:
        return False
    other_vals = [v for v in domains[row][col] if v != val]
    for v in other_vals:
        if not _eliminate(domains, row, col, v):
            return False
    return True


def _search_domains(domains: Domains) -> Optional[Domains]:
    # Check if solved
    if all(len(domains[r][c]) == 1 for r in range(9) for c in range(9)):
        return domains

    # Minimum Remaining Values (MRV) heuristic
    best_cell: Optional[Coord] = None
    best_len = 10
    for r in range(9):
        for c in range(9):
            d_len = len(domains[r][c])
            if 1 < d_len < best_len:
                best_len = d_len
                best_cell = (r, c)

    if best_cell is None:
        return None

    r, c = best_cell
    for val in sorted(domains[r][c]):
        next_domains = deepcopy(domains)
        if _assign(next_domains, r, c, val):
            solved = _search_domains(next_domains)
            if solved is not None:
                return solved

    return None


def solve_constraint_propagation(grid: Grid) -> Optional[Grid]:
    """
    Solves a 9x9 Sudoku grid using constraint propagation
    (naked singles, hidden singles, peer elimination) + MRV search.
    """
    if not is_standard_sudoku(grid):
        return None

    domains: Domains = [[set(range(1, 10)) for _ in range(9)] for _ in range(9)]

    for r in range(9):
        for c in range(9):
            val = grid[r][c]
            if val != 0:
                if not _assign(domains, r, c, val):
                    return None

    solved_domains = _search_domains(domains)
    if solved_domains is None:
        return None

    return [
        [next(iter(solved_domains[r][c])) for c in range(9)] for r in range(9)
    ]


# ==========================================
# Helper & Example Execution
# ==========================================
def print_grid(grid: Optional[Grid], title: str) -> None:
    print(f"\n--- {title} ---")
    if grid is None:
        print("No solution / Invalid puzzle.")
        return
    for r in range(9):
        if r > 0 and r % 3 == 0:
            print("-" * 21)
        row_str = ""
        for c in range(9):
            if c > 0 and c % 3 == 0:
                row_str += "| "
            val = grid[r][c]
            row_str += f"{val if val != 0 else '.'} "
        print(row_str.strip())


if __name__ == "__main__":
    # Placeholder 9x9 grid: verify/update cells to match your exact PDF layout
    puzzle_grid: Grid = [
        [5, 3, 0, 0, 7, 0, 0, 0, 0],
        [6, 0, 0, 1, 9, 5, 0, 0, 0],
        [0, 9, 8, 0, 0, 0, 0, 6, 0],
        [8, 0, 0, 0, 6, 0, 0, 0, 3],
        [4, 0, 0, 8, 0, 3, 0, 0, 1],
        [7, 0, 0, 0, 2, 0, 0, 0, 6],
        [0, 6, 0, 0, 0, 0, 2, 8, 0],
        [0, 0, 0, 4, 1, 9, 0, 0, 5],
        [0, 0, 0, 0, 8, 0, 0, 7, 9],
    ]

    print("Is standard Sudoku:", is_standard_sudoku(puzzle_grid))
    print_grid(puzzle_grid, "Initial Grid")

    bt_solution = solve_backtracking(puzzle_grid)
    print_grid(bt_solution, "Solved via Backtracking")

    cp_solution = solve_constraint_propagation(puzzle_grid)
    print_grid(cp_solution, "Solved via Constraint Propagation")
```

```python
from copy import deepcopy
from dataclasses import dataclass
from typing import Dict, List, Optional, Set, Tuple

Grid = List[List[int]]
Coord = Tuple[int, int]
Domains = List[List[Set[int]]]


@dataclass
class SolverResult:
    solution: Optional[Grid]
    steps: int
    branches: int = 0


# ==========================================
# 1. Check if the Sudoku is a Standard Puzzle
# ==========================================
def is_standard_sudoku(grid: Grid) -> bool:
    """
    Validates that `grid` is a standard 9x9 Sudoku board:
    - 9 rows and 9 columns
    - Integers in range 0..9 (0 = empty, 1..9 = given clue)
    - No duplicate non-zero digits in any row, column, or 3x3 box
    """
    if not isinstance(grid, list) or len(grid) != 9:
        return False

    for row in grid:
        if not isinstance(row, list) or len(row) != 9:
            return False
        for val in row:
            if type(val) is not int or val < 0 or val > 9:
                return False

    for i in range(9):
        row_seen: Set[int] = set()
        col_seen: Set[int] = set()
        for j in range(9):
            r_val = grid[i][j]
            c_val = grid[j][i]

            if r_val != 0:
                if r_val in row_seen:
                    return False
                row_seen.add(r_val)

            if c_val != 0:
                if c_val in col_seen:
                    return False
                col_seen.add(c_val)

    for box_r in range(0, 9, 3):
        for box_c in range(0, 9, 3):
            box_seen: Set[int] = set()
            for r in range(box_r, box_r + 3):
                for c in range(box_c, box_c + 3):
                    val = grid[r][c]
                    if val != 0:
                        if val in box_seen:
                            return False
                        box_seen.add(val)

    return True


def _build_peers_and_units() -> Tuple[Dict[Coord, Set[Coord]], Dict[Coord, List[List[Coord]]]]:
    peers_map: Dict[Coord, Set[Coord]] = {}
    units_map: Dict[Coord, List[List[Coord]]] = {}

    for r in range(9):
        for c in range(9):
            row_unit = [(r, cc) for cc in range(9)]
            col_unit = [(rr, c) for rr in range(9)]
            box_r, box_c = (r // 3) * 3, (c // 3) * 3
            box_unit = [
                (rr, cc) for rr in range(box_r, box_r + 3) for cc in range(box_c, box_c + 3)
            ]
            units_map[(r, c)] = [row_unit, col_unit, box_unit]

            peers: Set[Coord] = set()
            for unit in units_map[(r, c)]:
                for cell in unit:
                    if cell != (r, c):
                        peers.add(cell)
            peers_map[(r, c)] = peers

    return peers_map, units_map


PEERS_MAP, UNITS_MAP = _build_peers_and_units()


# ==========================================
# 2a. Optimized Backtracking Solver (with MRV + Counter)
# ==========================================
def _valid_candidates(grid: Grid, row: int, col: int) -> List[int]:
    used: Set[int] = set()
    for pr, pc in PEERS_MAP[(row, col)]:
        val = grid[pr][pc]
        if val != 0:
            used.add(val)
    return [num for num in range(1, 10) if num not in used]


def _select_mrv_empty(grid: Grid) -> Tuple[Optional[Coord], List[int]]:
    """Finds the empty cell with fewest valid candidates (MRV)."""
    best_cell: Optional[Coord] = None
    best_candidates: List[int] = []
    best_count = 10

    for r in range(9):
        for c in range(9):
            if grid[r][c] == 0:
                cands = _valid_candidates(grid, r, c)
                if len(cands) == 0:
                    return (r, c), []  # Immediate dead end
                if len(cands) < best_count:
                    best_count = len(cands)
                    best_cell = (r, c)
                    best_candidates = cands
                    if best_count == 1:
                        return best_cell, best_candidates

    return best_cell, best_candidates


def solve_backtracking(grid: Grid) -> SolverResult:
    """
    Solves a 9x9 Sudoku grid using backtracking with MRV ordering,
    tracking total placement steps and search branches.
    """
    if not is_standard_sudoku(grid):
        return SolverResult(solution=None, steps=0, branches=0)

    board = deepcopy(grid)
    steps = 0
    branches = 0

    def backtrack() -> bool:
        nonlocal steps, branches
        empty, candidates = _select_mrv_empty(board)
        if empty is None:
            return True
        if not candidates:
            return False

        row, col = empty
        for num in candidates:
            steps += 1
            branches += 1
            board[row][col] = num

            if backtrack():
                return True

            board[row][col] = 0  # Undo placement on backtrack

        return False

    solved = backtrack()
    return SolverResult(solution=board if solved else None, steps=steps, branches=branches)


# ==========================================
# 2b. Constraint Propagation + Search Solver (with Counter)
# ==========================================
def solve_constraint_propagation(grid: Grid) -> SolverResult:
    """
    Solves a 9x9 Sudoku grid using constraint propagation
    (naked singles, hidden singles, peer elimination) + MRV search,
    tracking total propagation/assignment steps and search branches.
    """
    if not is_standard_sudoku(grid):
        return SolverResult(solution=None, steps=0, branches=0)

    steps = 0
    branches = 0

    def eliminate(domains: Domains, row: int, col: int, val: int) -> bool:
        nonlocal steps
        if val not in domains[row][col]:
            return True

        steps += 1
        domains[row][col].remove(val)

        if len(domains[row][col]) == 0:
            return False

        # Naked single: only one candidate left in this cell
        if len(domains[row][col]) == 1:
            only_val = next(iter(domains[row][col]))
            for pr, pc in PEERS_MAP[(row, col)]:
                if not eliminate(domains, pr, pc, only_val):
                    return False

        # Hidden single: `val` has only one valid cell in a row, column, or box
        for unit in UNITS_MAP[(row, col)]:
            places = [(r, c) for (r, c) in unit if val in domains[r][c]]
            if len(places) == 0:
                return False
            if len(places) == 1:
                ur, uc = places[0]
                if len(domains[ur][uc]) > 1:
                    if not assign(domains, ur, uc, val):
                        return False

        return True

    def assign(domains: Domains, row: int, col: int, val: int) -> bool:
        nonlocal steps
        if val not in domains[row][col]:
            return False
        steps += 1
        other_vals = [v for v in domains[row][col] if v != val]
        for v in other_vals:
            if not eliminate(domains, row, col, v):
                return False
        return True

    def search(domains: Domains) -> Optional[Domains]:
        nonlocal branches
        if all(len(domains[r][c]) == 1 for r in range(9) for c in range(9)):
            return domains

        best_cell: Optional[Coord] = None
        best_len = 10
        for r in range(9):
            for c in range(9):
                d_len = len(domains[r][c])
                if 1 < d_len < best_len:
                    best_len = d_len
                    best_cell = (r, c)
                    if best_len == 2:
                        break
            if best_len == 2:
                break

        if best_cell is None:
            return None

        r, c = best_cell
        for val in sorted(domains[r][c]):
            branches += 1
            next_domains = deepcopy(domains)
            if assign(next_domains, r, c, val):
                solved = search(next_domains)
                if solved is not None:
                    return solved

        return None

    initial_domains: Domains = [[set(range(1, 10)) for _ in range(9)] for _ in range(9)]
    for r in range(9):
        for c in range(9):
            val = grid[r][c]
            if val != 0:
                if not assign(initial_domains, r, c, val):
                    return SolverResult(solution=None, steps=steps, branches=branches)

    solved_domains = search(initial_domains)
    if solved_domains is None:
        return SolverResult(solution=None, steps=steps, branches=branches)

    solution = [
        [next(iter(solved_domains[r][c])) for c in range(9)] for r in range(9)
    ]
    return SolverResult(solution=solution, steps=steps, branches=branches)


# ==========================================
# Helper & Comparison Execution
# ==========================================
def print_grid(grid: Optional[Grid], title: str) -> None:
    print(f"\n--- {title} ---")
    if grid is None:
        print("No solution / Invalid puzzle.")
        return
    for r in range(9):
        if r > 0 and r % 3 == 0:
            print("-" * 21)
        row_str = ""
        for c in range(9):
            if c > 0 and c % 3 == 0:
                row_str += "| "
            val = grid[r][c]
            row_str += f"{val if val != 0 else '.'} "
        print(row_str.strip())


if __name__ == "__main__":
    # Placeholder 9x9 grid: verify/update cells to match your exact PDF layout
    puzzle_grid: Grid = [
        [5, 3, 0, 0, 7, 0, 0, 0, 0],
        [6, 0, 0, 1, 9, 5, 0, 0, 0],
        [0, 9, 8, 0, 0, 0, 0, 6, 0],
        [8, 0, 0, 0, 6, 0, 0, 0, 3],
        [4, 0, 0, 8, 0, 3, 0, 0, 1],
        [7, 0, 0, 0, 2, 0, 0, 0, 6],
        [0, 6, 0, 0, 0, 0, 2, 8, 0],
        [0, 0, 0, 4, 1, 9, 0, 0, 5],
        [0, 0, 0, 0, 8, 0, 0, 7, 9],
    ]

    print("Is standard Sudoku:", is_standard_sudoku(puzzle_grid))
    print_grid(puzzle_grid, "Initial Grid")

    bt_res = solve_backtracking(puzzle_grid)
    print_grid(bt_res.solution, f"Backtracking (steps={bt_res.steps}, branches={bt_res.branches})")

    cp_res = solve_constraint_propagation(puzzle_grid)
    print_grid(
        cp_res.solution,
        f"Constraint Propagation (steps={cp_res.steps}, branches={cp_res.branches})",
    )
```

```python
"""
Sudoku Assignment (Simplified Version)
1. Check if a puzzle is a standard Sudoku
2. Solve it with backtracking and constraint propagation
"""

import time

# The puzzle we have to solve (0 = empty cell)
PUZZLE = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9],
]


# Helper: check if putting number 'num' in grid[row][col] is legal
def is_ok(grid, row, col, num):
    # check the row
    for c in range(9):
        if grid[row][c] == num:
            return False
    # check the column
    for r in range(9):
        if grid[r][col] == num:
            return False
    # check the 3x3 box
    box_row = row - row % 3
    box_col = col - col % 3
    for r in range(box_row, box_row + 3):
        for c in range(box_col, box_col + 3):
            if grid[r][c] == num:
                return False
    return True


# Helper: find an empty cell, returns (row, col) or None if board is full
def find_empty(grid):
    for r in range(9):
        for c in range(9):
            if grid[r][c] == 0:
                return r, c
    return None


# Check if the final grid is a valid solved sudoku
def is_valid_solution(grid):
    for r in range(9):
        for c in range(9):
            num = grid[r][c]
            if num == 0:
                return False
            # temporarily remove the number, then check if it's legal
            grid[r][c] = 0
            ok = is_ok(grid, r, c, num)
            grid[r][c] = num
            if not ok:
                return False
    return True


# ---------------- Task 1: check if it's a standard sudoku ----------------
def count_solutions(grid, stop_at=2):
    """Count solutions. We stop early once we reach 'stop_at' solutions."""
    # find an empty cell
    empty = find_empty(grid)
    if empty is None:
        return 1  # board is full, we found one solution

    row, col = empty
    total = 0
    for num in range(1, 10):
        if is_ok(grid, row, col, num):
            grid[row][col] = num          # try the number
            total += count_solutions(grid, stop_at)
            grid[row][col] = 0            # undo (backtrack)
            if total >= stop_at:          # no need to count further
                return total
    return total


def check_standard_sudoku(grid):
    """Returns (True/False, reason)."""
    # 1. check the shape
    if len(grid) != 9:
        return False, "grid is not 9 x 9"
    for row in grid:
        if len(row) != 9:
            return False, "grid is not 9 x 9"

    # 2. check the values
    for row in grid:
        for v in row:
            if not isinstance(v, int) or v < 0 or v > 9:
                return False, "invalid value " + str(v)

    # 3. check no number repeats in any row, column or box
    for r in range(9):
        for c in range(9):
            if grid[r][c] != 0 and not is_ok(grid, r, c, grid[r][c]):
                return False, "a digit repeats in a row, column or box"

    # 4. check there is exactly one solution
    copy = [row[:] for row in grid]   # work on a copy so we don't change the original
    n = count_solutions(copy)
    if n == 0:
        return False, "no solution"
    if n > 1:
        return False, "more than one solution"

    return True, "9 x 9, no conflicts, exactly one solution"


# ---------------- Task 2a: solve with backtracking ----------------
def solve_backtracking(grid):
    """Returns (solution or None, calls, backtracks)."""
    empty = find_empty(grid)
    if empty is None:
        return True, 1, 0   # no empty cells -> solved

    row, col = empty
    calls = 1
    backtracks = 0
    for num in range(1, 10):
        if is_ok(grid, row, col, num):
            grid[row][col] = num
            solved, extra_calls, extra_backtracks = solve_backtracking(grid)
            calls += extra_calls
            backtracks += extra_backtracks
            if solved:
                return True, calls, backtracks
            grid[row][col] = 0    # undo
            backtracks += 1
    return False, calls, backtracks


# ---------------- Task 2b: solve with constraint propagation ----------------
def solve_constraint_propagation(grid):
    """
    Simple version: fill in every cell that has only ONE possible number,
    and repeat until nothing changes. If the puzzle isn't solved after that,
    we fall back to guessing with backtracking.
    """
    guesses = 0

    changed = True
    while changed:
        changed = False
        for r in range(9):
            for c in range(9):
                if grid[r][c] == 0:
                    possible = []
                    for num in range(1, 10):
                        if is_ok(grid, r, c, num):
                            possible.append(num)
                    if len(possible) == 0:
                        return None, guesses   # contradiction, no solution
                    if len(possible) == 1:
                        grid[r][c] = possible[0]   # only one choice -> fill it in
                        changed = True

    # if there are still empty cells, we have to guess
    empty = find_empty(grid)
    if empty is not None:
        row, col = empty
        for num in range(1, 10):
            if is_ok(grid, row, col, num):
                guesses += 1
                grid[row][col] = num
                copy = [r2[:] for r2 in grid]
                result, extra = solve_constraint_propagation(copy)
                guesses += extra
                if result is not None:
                    for r in range(9):
                        for c in range(9):
                            grid[r][c] = result[r][c]
                    return grid, guesses
                grid[row][col] = 0
        return None, guesses

    return grid, guesses


# ---------------- Main ----------------
if __name__ == "__main__":
    print("Task 1:", check_standard_sudoku(PUZZLE))

    # backtracking
    grid1 = [row[:] for row in PUZZLE]
    t = time.perf_counter()
    solved1, calls, backtracks = solve_backtracking(grid1)
    t_bt = (time.perf_counter() - t) * 1000
    print("\nBacktracking solution:")
    for row in solved1:
        print(row)
    print("calls =", calls, ", backtracks =", backtracks, ", time =", round(t_bt, 2), "ms")

    # constraint propagation
    grid2 = [row[:] for row in PUZZLE]
    t = time.perf_counter()
    solved2, guesses = solve_constraint_propagation(grid2)
    t_cp = (time.perf_counter() - t) * 1000
    print("\nConstraint propagation solution:")
    for row in solved2:
        print(row)
    print("guesses =", guesses, ", time =", round(t_cp, 2), "ms")

    print("\nSame solution:", solved1 == solved2, "| Valid:", is_valid_solution(solved2))
```
